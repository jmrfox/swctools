```python
import logging
logging.basicConfig(level=logging.DEBUG)

from pathlib import Path
from swctools import FrustaSet, plot_frusta, animate_frusta_timeseries

swc_filepath = Path("../data/swc/TS2_s50.swc")
frusta = FrustaSet.from_swc_file(swc_filepath)
fig_frusta = plot_frusta(frusta)
fig_frusta.show()
```

```python
# create a timeseries data block and animate it using animate_frusta_timeseries
import numpy as np

n_frusta = frusta.n_frusta
T = 10
dt = 0.1
time_domain = np.linspace(0, T, int(T / dt) + 1)
freqs = np.random.uniform(3, 30, n_frusta) / T
V = np.zeros((len(time_domain), n_frusta))
for i, t in enumerate(time_domain):
    V[i, :] = np.sin(freqs * t)

fig = animate_frusta_timeseries(
    frusta,
    time_domain=time_domain,
    amplitudes=V,
    fps=10,
)
fig.show()
```
