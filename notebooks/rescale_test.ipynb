```python

from pathlib import Path
from swctools import SWCModel, plot_model

swc_px_path = Path("../data/swc/TS2_s50.swc")
swc_um_path = Path("../data/swc/TS2_s50_um.swc")

model_px = SWCModel.from_swc_file(swc_px_path)
plot_model(swc_model=model_px)
```

```python
model_um = model_px.scale(5/1000)
model_um.to_swc_file(swc_um_path)
plot_model(swc_model=model_um)
```

```python
model_um = SWCModel.from_swc_file(swc_um_path)
plot_model(swc_model=model_um)
```
